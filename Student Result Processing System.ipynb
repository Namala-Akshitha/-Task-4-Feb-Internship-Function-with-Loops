```python
def check_result(marks):
    total = 0
    #Using loop to calculate total
    for mark in marks:
        total += mark
    average = total / len(marks)
    print("Average Marks:", average)
    if average >= 50:
        print("Result: Pass")
        return "Pass"
    else:
        print("Result: Fail ❌")
        return "Fail"
#Example usage
student_marks = [60, 45, 70, 55]
result = check_result(student_marks)
```
