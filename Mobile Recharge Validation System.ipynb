```python
def mobile_recharge(amount):
    valid_plans = [199, 299, 399, 599]
    #Minimum recharge amount
    if amount < 50:
        print("Recharge amount must be at least ₹50.")
        return False
    #Must match valid plans
    if amount not in valid_plans:
        print("Invalid Plan! Available plans are:", valid_plans)
        return False
    #If all conditions are satisfied
    print("Recharge Successful for ₹", amount)
    return True
#Main Program with Retry Option
while True:
    amt = int(input("Enter recharge amount: "))
    if mobile_recharge(amt):
        break   #Exit loop if recharge is valid
    else:
        print("Please try again...\n")
```
