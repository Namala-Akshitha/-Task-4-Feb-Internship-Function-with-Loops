```python
def check_stock(products):
    for item,stock in products.items():
        if stock < 15:
            print(item, ": Reorder Alert")
        else:
            print(item, ": Stock OK")
#Dictionary of products and stock quantities
product_stock = {
    "Rice": 10,
    "Sugar": 25,
    "Oil": 12,
    "Salt": 30,
    "Tea": 8
}
#Function Call
check_stock(product_stock)
```
