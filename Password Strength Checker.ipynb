```python
def check_password(password):
    #Minimum length
    if len(password) < 8:
        return "Weak Password (Minimum 8 characters required)"
    has_digit = False
    has_special = False
    #Loop over string
    for char in password:
        if char.isdigit():
            has_digit=True
        if char in "@#$":
            has_special=True
    #Conditions check
    if has_digit and has_special:
        return "Strong Password"
    else:
        return "Weak Password (Must contain at least 1 digit and 1 special character @#$)"
#Example
pwd=input("Enter password: ")
result=check_password(pwd)
print(result)
```
