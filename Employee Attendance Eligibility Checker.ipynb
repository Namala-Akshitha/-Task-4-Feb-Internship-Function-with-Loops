```python
def attendance_eligibility(attendance_list):
    count=0
    #Loop to count days
    for day in attendance_list:
        if day=="P":
            count+=1
    total_days=len(attendance_list)
    percentage=(count/total_days)*100
    print("Attendance Percentage:", percentage, "%")
    #return result
    if percentage >= 75:
        return "Eligible"
    else:
        return "Not Eligible"
attendance=["P", "A", "P", "P", "A", "P", "P", "P"]
result=attendance_eligibility(attendance)
print("Result:", result)
```
